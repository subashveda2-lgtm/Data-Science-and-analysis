```python
#--------corelation between hours and math--------------
corr_matrix = np.corrcoef(hours, math)
correlation = corr_matrix[0, 1] 
print(f"Correlation between hours studied and math Scores: {correlation:.4f}")
print()

if correlation > 0.5:
    print("Strong positive link- Studying more = better math!") 
elif correlation > 0.2: 
    print("Weak positive link - slight trend, but not strong.") 
else:
    print("Very weak/no link - hours studied alone may not predict score.")
 
```

```python
df["Total Score"] = total_score
df["Average Score"] = np.round(student_avg, 2)
df["Rank"] = ranks = np.argsort(np.argsort(-student_avg)) + 1
df["GPA"] = np.round(student_avg / 90 * 4,1)
df["Result"] = np.where(student_avg >= 50, "Pass", "Fail")

# Assign letter grades
def assign_grade(avg):
    if avg >= 90: return "A+"
    elif avg >= 80: return "A"
    elif avg >= 70: return "B+"
    elif avg >= 60: return "B"
    elif avg >= 50: return "C+"
    elif avg >= 40: return "C"
    elif avg >= 30: return "D+"
    elif avg >= 20: return "D"
    else: return "F"

df["Grade"] = [assign_grade(a) for a in student_avg]

def assign_division(avg):
    if avg >= 80: return "Distinction"
    elif avg >= 60: return "First Division"
    elif avg >= 50: return "Second Division"
    elif avg >=40: return "Third Division"
    else: return "Fail"
    
df["Division"] = [assign_division(a) for a in student_avg]

# Save to a new CSV
df.to_csv("student_results.csv", index=False)

print("Final result saved to student_results.csv!")
print()
print("Grade Distribution:")
print(df["Grade"].value_counts().sort_index())
print()

print("First 5 rows of final data with all columns:")
df.head()
```
