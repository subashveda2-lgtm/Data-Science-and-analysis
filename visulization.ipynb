```python
gender = df.groupby("Gender")["AttendancePercent"].mean()
plt.figure(figsize=(10, 5)) 
plt.plot(gender.index, gender.values,  marker="o" )
plt.title("Average Attendance Percent by Gender") 
plt.xlabel("Gender") 
plt.ylabel("Average Attendance Percent") 
plt.grid(True)
plt.show()
```

```python
studyhours = df.groupby("StudyHours")["Passed"].sum()
plt.figure(figsize=(10, 5))
plt.bar(studyhours.index, studyhours.values, color= "blue") 
plt.title("Total Passes Students by Study hours") 
plt.xlabel("Study Hours") 
plt.ylabel("Total Passes Students")
plt.show()
```

```python
studyhours = df.groupby("StudyHours")["Rank"].mean()
plt.figure(figsize=(10, 5))
plt.bar(studyhours.index, studyhours.values, color= "green") 
plt.title("Total Rank by study hours") 
plt.xlabel("Study Hours") 
plt.ylabel("Total Rank") 
plt.show()
```

```python
studyhours = df.groupby("StudyHours")["Rank"].mean()
plt.figure(figsize=(10, 5))
plt.plot(studyhours.index, studyhours.values, marker = "s") 
plt.title("Total Rank by study hours") 
plt.xlabel("Study Hours") 
plt.ylabel("Total Rank") 
plt.grid(True)
plt.show()
```

```python
parentaleducation = df.groupby("ParentalEducation")["Passed"].sum()
plt.figure(figsize=(10, 5)) 
plt.pie(parentaleducation.values, labels=parentaleducation.index, autopct = "%1.1f%%") 
plt.title("Total passed students by parental education")
plt.show()
```

```python
parentaleducation = df.groupby("ParentalEducation")["Passed"].sum()
plt.figure(figsize=(10, 5)) 
plt.scatter(parentaleducation.index, parentaleducation.values, alpha=0.7)
plt.xlabel("Parental Education")
plt.ylabel("Total Passed Students")
plt.title("Total passed students by parental education")
plt.grid(True)
plt.show()
```

```python
grade = df.groupby("Grade")["Passed"].mean()
plt.figure(figsize= (10, 5)) 
plt.bar(grade.index, grade.values, color="blue")
plt.title("Average Grade by pass status") 
plt.xlabel("Grade") 
plt.ylabel("Average Passed Students") 
plt.show()
```

```python
plt.figure(figsize=(10, 5)) 
plt.scatter(df["StudyHours"], df["Passed"], color="purple", marker="o")
plt.title("Correlation between Study Hours and pass status") 
plt.xlabel("Study Hours") 
plt.ylabel("Pass Status") 
plt.show()
```

```python
ct_gender_passed = pd.crosstab(df["Gender"], df["Passed"])
plt.figure(figsize=(10, 5)) 
sns.heatmap(ct_gender_passed, annot=True, fmt="d", cmap="YlGnBu")
plt.title("Cross Tabulation between gender and Passed Status")
plt.xlabel("Passed Status") 
plt.ylabel("Gender") 
plt.show()
```

```python
#----------Data Analysis Dashboard---------------
total_data = len(df)
avg_score = df["Average Score"].mean()
total_passed = df["Passed"].sum()
total_failed = len(df) - total_passed

fig,axes = plt.subplots(3, 2, figsize=(15, 10))
fig.suptitle("Data Analysis of Student Performance", fontsize=16, fontweight="bold", y=0.95)


#--chart 1
gender = df.groupby("Gender")["Passed"].mean()
axes[0, 0].plot(gender.index, gender.values, color="blue", marker="o")
axes[0, 0].set_title("Average Pass", fontsize=15, fontweight="bold")
axes[0, 0].set_ylabel("Average Pass")
axes[0, 0].set_xlabel("Gender")
axes[0, 0].grid(True)
#-----chart 2     
parentaleducation = df.groupby("ParentalEducation")["Passed"].mean()
axes[0, 1].bar(parentaleducation.index, parentaleducation.values, color="green")
axes[0, 1].set_title("Average Pass by Parental Education", fontsize=15, fontweight="bold")
axes[0, 1].set_ylabel("Parental Education")
axes[0, 1].set_xlabel("Average Pass")


#chart 3
studyhours = df.groupby("StudyHours")["Passed"].mean()
axes[1, 0].scatter(studyhours.index, studyhours.values, color="purple", marker="o")
axes[1, 0].set_title("Average Pass by Study Hours", fontsize=15, fontweight="bold")
axes[1, 0].set_ylabel("Average Pass")
axes[1, 0].set_xlabel("Study Hours")  
axes[1,0].grid(True)

#chart 4
extracurricular = df.groupby("ExtraCurricular")["Passed"].mean()
axes[1, 1].pie(extracurricular.values, labels=extracurricular.index, autopct="%1.1f%%")
axes[1, 1].set_title("Average Pass by Extra-Curricular Activities", fontsize=15, fontweight="bold")
axes[1, 1].set_ylabel("Extra-Curricular Activities")
axes[1, 1].set_xlabel("Average Pass")


#chart 5
attendancepercent= df.groupby("AttendancePercent")["Passed"].mean()
axes[2, 0].bar(attendancepercent.index, attendancepercent.values, color="green")
axes[2, 0].set_title("Average Pass by Attendance Percent", fontsize=15, fontweight="bold")
axes[2, 0].set_ylabel("Attendance Percent")
axes[2, 0].set_xlabel("Passed Status")


#chart 6
internetaccess = df.groupby("InternetAccess")["Passed"].mean()
axes[2, 1].scatter(internetaccess.index, internetaccess.values, color="blue", marker="s")
axes[2, 1].set_title("Average Pass by Internet Access", fontsize=15, fontweight="bold")
axes[2, 1].set_ylabel("Internet Access")
axes[2, 1].set_xlabel("Average Pass")
axes[2, 1].grid(True)

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()

#-----------Text based Summary-------------
print("-"*50)
print("Student Performance Data Analysis Summary")
print(f"Total Students: {total_data}")
print(f"Average Score: {avg_score:.2f}")
print(f"Total Passed: {total_passed}")
print(f"Total Failed: {total_failed}")
print("-"*50)
print("critical insights:")
print("1. Students with higher study hours tend to have a higher pass rate.")
print("2. Parental education level has a significant impact on student performance.")
print("3. Internet access is positively correlated with better academic outcomes.")
print("4. Extra-curricular activities contribute to overall student success.")
print("5. Gender differences in pass rates may indicate the need for targeted support and interventions.")
print("6. Grade level is a strong predictor of student performance, with higher grades associated with higher pass rates.")
```
