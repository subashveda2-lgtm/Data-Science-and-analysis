```python
print("Attendance Percent Average")
print("-" * 35)
print(f"mean of attendance percent: {np.mean(attendancepercent):.2f}") 
print(f"median : {np.median(attendancepercent):.2f}")
print(f"standard deviation: {np.std(attendancepercent):.2f}") 
print(f"Minimum: {np.min(attendancepercent)}") 
print(f"Maximum: {np.max(attendancepercent)}") 
print(f"Total sum: {np.sum(attendancepercent)}") 
print()

print("Previous Score Average")
print("-" * 35)
print(f"mean of previous score: {np.mean(previousscore):.2f}") 
print(f"median : {np.median(previousscore):.2f}")
print(f"standard deviation: {np.std(previousscore):.2f}") 
print(f"Minimum: {np.min(previousscore)}") 
print(f"Maximum: {np.max(previousscore)}") 
print(f"Total sum: {np.sum(previousscore)}") 
print()

print("Final Score Average")
print("-" * 35)
print(f"mean of final score: {np.mean(finalscore):.2f}") 
print(f"median : {np.median(finalscore):.2f}")
print(f"standard deviation: {np.std(finalscore):.2f}") 
print(f"Minimum: {np.min(finalscore)}") 
print(f"Maximum: {np.max(finalscore)}") 
print(f"Total sum: {np.sum(finalscore)}") 
print()

```

```python
scores_2d = np.column_stack((attendancepercent, previousscore, finalscore)) 
print("2D Array of scores :", scores_2d.shape) 
print("500 rows (students), 3 columns (subjects)") 

print("First 5 rows (students):") 
print("Attendance | Previous | Final") 
print(scores_2d[:5]) 

print()

#Average score across all subjects for each student-----------
subject_avg = np.mean(scores_2d, axis=0) 
print(f"Average by subjects") 
print(f"Attendance : {subject_avg[0]:.2f}") 
print(f"Previous : {subject_avg[1]:.2f}")
print(f"Final : {subject_avg[2]:.2f}") 

print() 

student_avg =np.mean(scores_2d, axis=1) 
print("Average Score for each student (first 5):", np.round(student_avg[:5], 2)) 

```

```python
data["Total Score"] = total_score
data["Average Score"] = np.round(student_avg, 2)
data["Rank"] = ranks = np.argsort(np.argsort(-student_avg)) + 1
data["GPA"] = np.round(student_avg / 90 * 4,1)
# Assign letter grades
def assign_grade(avg):
    if avg >= 90: return "A+"
    elif avg >= 80: return "A"
    elif avg >= 70: return "B+"
    elif avg >= 60: return "B"
    elif avg >= 50: return "C+"
    elif avg >= 40: return "C"
    elif avg >= 30: return "D+"
    elif avg >= 20: return "D"
    else: return "F"

data["Grade"] = [assign_grade(a) for a in student_avg]

def assign_division(avg):
    if avg >= 80: return "Distinction"
    elif avg >= 60: return "First Division"
    elif avg >= 50: return "Second Division"
    elif avg >=40: return "Third Division"
    else: return "Fail"
    
data["Division"] = [assign_division(a) for a in student_avg]

# Save to a new CSV
data.to_csv("studentproject_results.csv", index=False)

print("Final result saved to studentproject_results.csv!")
print()
print("Grade Distribution:")
print(data["Grade"].value_counts().sort_index())
print()

print("First 5 rows of final data with all columns:")
data.head()
```
