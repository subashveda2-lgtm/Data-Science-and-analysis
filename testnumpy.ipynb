```python
# show the average score of all subjects-----------
print("Math Average Scores") 
print("-" * 35) 
print(f"mean of math: {np.mean(math):.2f}")
print(f"median of math: {np.median(math):.2f}") 
print(f"standard deviation: {np.std(math):.2f}") 
print(f"minimum of math: {np.min(math)}") 
print(f"maximum of math: {np.max(math)}") 
print(f"total sum of math: {np.sum(math)}")

print() 

print("Science Average Scores") 
print("-" * 35) 
print(f"mean of science: {np.mean(science):.2f}")
print(f"median of science: {np.median(science):.2f}") 
print(f"standard deviation: {np.std(science):.2f}") 
print(f"minimum of science: {np.min(science)}") 
print(f"maximum of science: {np.max(science)}") 
print(f"total sum of science: {np.sum(science)}")

print()

print("English Average Scores") 
print("-" * 35) 
print(f"mean of english: {np.mean(english):.2f}")
print(f"median of english: {np.median(english):.2f}") 
print(f"standard deviation: {np.std(english):.2f}") 
print(f"minimum of english: {np.min(english)}") 
print(f"maximum of english: {np.max(english)}") 
print(f"total sum of english: {np.sum(english)}")

print()
```

```python
scores_2d = np.column_stack((math, science, english)) 
print("2D Array of scores :", scores_2d.shape) 
print("500 rows (students), 3 columns (subjects)") 

print("First 5 rows (students):") 
print("Math | Science | English") 
print(scores_2d[:5]) 

print()

#Average score across all subjects for each student-----------
subject_avg = np.mean(scores_2d, axis=0) 
print(f"Average by subjects") 
print(f"Math : {subject_avg[0]:.2f}") 
print(f"Science : {subject_avg[1]:.2f}")
print(f"English : {subject_avg[2]:.2f}") 

print() 

student_avg =np.mean(scores_2d, axis=1) 
print("Average Score for each student (first 5):", np.round(student_avg[:5], 2)) 

```

```python
test["Total Score"] = total_score
test["Average Score"] = np.round(student_avg, 2)
test["Rank"] = ranks = np.argsort(np.argsort(-student_avg)) + 1
test["GPA"] = np.round(student_avg / 90 * 4,1)
test["Result"] = np.where(student_avg >= 50, "Pass", "Fail")

# Assign letter grades
def assign_grade(avg):
    if avg >= 90: return "A+"
    elif avg >= 80: return "A"
    elif avg >= 70: return "B+"
    elif avg >= 60: return "B"
    elif avg >= 50: return "C+"
    elif avg >= 40: return "C"
    elif avg >= 30: return "D+"
    elif avg >= 20: return "D"
    else: return "F"

test["Grade"] = [assign_grade(a) for a in student_avg]

def assign_division(avg):
    if avg >= 80: return "Distinction"
    elif avg >= 60: return "First Division"
    elif avg >= 50: return "Second Division"
    elif avg >=40: return "Third Division"
    else: return "Fail"
    
test["Division"] = [assign_division(a) for a in student_avg]

# Save to a new CSV
test.to_csv("studenttest_results.csv", index=False)

print("Final result saved to studenttest_results.csv!")
print()
print("Grade Distribution:")
print(test["Grade"].value_counts().sort_index())
print()

print("First 5 rows of final data with all columns:")
test.head()


```
